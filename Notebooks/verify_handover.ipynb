```python
#for each customer,how many orders did they place?,
#how many of these were 500 or more?
q("""SELECT COUNT(*) AS orders,
SUM(amount>=500) as big_orders,
ROUND(100.0*avg(amount>=500),2) as perc
FROM shop_orders;""")
```

```python
# On 31 March 2025, how any days had each customer been with the shop?
# customer name ,jioned_date,no of days
q("""SELECT name,
joined_on,
CAST(julianday('2025-03-31') - julianday(joined_on) AS INT) AS days
FROM shop_customers;
""")
```

```python
#How many orders and how much sales in each month ?
#month ,order, sales 
#shop_orders

q("""SELECT strftime('%Y-%m', order_date) AS month,   
       COUNT(*)    AS orders,                  
       SUM(amount) AS sales                      
FROM   shop_orders                           
GROUP  BY month ORDER BY month; """)
```

```python
# which customers have never placed an order?

q("""SELECT c.name,c.city
FROM shop_customers AS c
LEFT JOIN shop_orders AS o
    on o.customer_id=c.customer_id
WHERE o.customer_id IS NULL;""") 
```

```python
# Which member have never borrowed a book?
q("""SELECT m.member_id, m.name
FROM members AS m
LEFT JOIN loans AS l
    ON l.member_id = m.member_id
WHERE l.member_id IS NULL
ORDER BY m.member_id;""")
```
